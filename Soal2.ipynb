```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img_jalan = cv2.imread('jalan.jpg',0)
rows, cols = img_jalan.shape
print("IMG Shape: ", img_jalan.shape)
```

```python
res = cv2.resize(img_jalan, None, fx=9, fy=5, interpolation=cv2.INTER_CUBIC)
```

```python
M = cv2.getRotationMatrix2D(((cols-1)/2.0, (rows-1)/2.0), -90,1)
img_putar = cv2.warpAffine(img_jalan, M, (cols,rows))

fig, axs = plt.subplots(1,2,figsize = (10,5))
axs = axs.ravel()

axs[0].imshow(img_jalan, cmap='gray')
axs[0].set_title("Gambar Asli")

axs[1].imshow(img_putar, cmap='gray')
axs[1].set_title("Resized IMG")

for a in axs:
    a.axis('off')

plt.tight_layout()
plt.show()
```

```python
from skimage import io, transform

img_jalan2 = io.imread('jalan.jpg')

rotated = transform.rotate(img_jalan2, 45, resize=False)
rotated2 = transform.rotate(img_jalan2, 45, resize=True)

fig, axs = plt.subplots(1,3,figsize = (10,5))
axs = axs.ravel()

axs[0].imshow(img_jalan2, cmap='gray')
axs[0].set_title("Gambar Asli")

axs[1].imshow(rotated, cmap='gray')
axs[1].set_title("Gambar Putaran 1")

axs[2].imshow(rotated2, cmap='gray')
axs[2].set_title("Gambar Putaran 2")

for a in axs:
    a.axis('off')

plt.tight_layout()
plt.show()
```
