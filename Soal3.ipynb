```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
t = cv2.imread('quotes.jpeg')
plt.imshow(t)
```

```python
from skimage import io, transform
src = np.array([
    [0,0],
    [0,50],
    [300,50],
    [300,0]
])

crp = np.array([
    [74,397],
    [94,462],
    [677,353],
    [655,297]
])

crp2 = np.array([
    [184,469],
    [204,533],
    [559,476],
    [551,409]
])



tform = transform.ProjectiveTransform()
tform.estimate(src, crp)

tform2 = transform.ProjectiveTransform()
tform2.estimate(src, crp2)

warped = transform.warp(t, tform, output_shape=(50,300))
warped2 = transform.warp(t, tform2, output_shape=(50,300))

fig, axs = plt.subplots(1,3,figsize = (10,5))
axs = axs.ravel()

axs[0].imshow(warped)
axs[0].set_title("Warped IMG 1")

axs[1].imshow(t)
axs[1].plot(crp[:,0], crp[:,1], '.r')
axs[1].plot(crp2[:,0], crp2[:,1], '.b')
axs[1].set_title("Gambar Ori")

axs[2].imshow(warped2)
axs[2].set_title("Warped IMG 2")

for a in axs:
    a.axis('off')

plt.tight_layout()
plt.show()
```
